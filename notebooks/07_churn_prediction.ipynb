```python
customer_df = pd.read_csv(
    "../data/customer_details.csv"
)

customer_df.head()
```

```python
customer_df["Last_Purchase_Date"] = pd.to_datetime(
    customer_df["Last_Purchase_Date"]
)
```

```python
import pandas as pd

today = pd.Timestamp.today()

customer_df["Days_Since_Last_Purchase"] = (
    today
    - customer_df["Last_Purchase_Date"]
).dt.days
```

```python
customer_df["Churn_Risk"] = "Low"

customer_df.loc[
    customer_df["Days_Since_Last_Purchase"] > 180,
    "Churn_Risk"
] = "High"

customer_df.loc[
    (
        customer_df["Days_Since_Last_Purchase"] > 90
    )
    &
    (
        customer_df["Days_Since_Last_Purchase"] <= 180
    ),
    "Churn_Risk"
] = "Medium"
```

```python
customer_df.loc[
    (
        customer_df["Loyalty_Status"]
        == "Gold"
    )
    &
    (
        customer_df["Churn_Risk"]
        == "Medium"
    ),
    "Churn_Risk"
] = "Low"
```

```python
customer_df["Churn_Risk"].value_counts()
```

```python
customer_df.to_csv(
    "../outputs/churn_predictions.csv",
    index=False
)
```
