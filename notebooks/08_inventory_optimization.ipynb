```python
import pandas as pd

df = pd.read_csv(
    "../data/dataset_3_sku_level_inventory_forecasting.csv"
)

df.head()
```

```python
print(df.columns.tolist())
```

```python
inventory_df = df.copy()

inventory_df["Safety_Stock"] = (
    inventory_df["Average_Daily_Demand"] * 7
)

inventory_df["Reorder_Point"] = (
    inventory_df["Average_Daily_Demand"] * 14
)

inventory_df["Inventory_Gap"] = (
    inventory_df["Reorder_Point"]
    - inventory_df["Net_Quantity"]
)

inventory_df.head()
```

```python
inventory_df["Stock_Status"] = "Optimal"

inventory_df.loc[
    inventory_df["Net_Quantity"]
    < inventory_df["Reorder_Point"],
    "Stock_Status"
] = "Reorder Required"

inventory_df.loc[
    inventory_df["Net_Quantity"]
    >
    inventory_df["Reorder_Point"] * 1.5,
    "Stock_Status"
] = "Overstock"

inventory_df["Stock_Status"].value_counts()
```

```python
inventory_df["Recommendation"] = (
    inventory_df["Stock_Status"]
)

inventory_df.loc[
    inventory_df["Stock_Status"]
    == "Reorder Required",
    "Recommendation"
] = "Reorder Immediately"

inventory_df.loc[
    inventory_df["Stock_Status"]
    == "Optimal",
    "Recommendation"
] = "Maintain Current Stock"

inventory_df.loc[
    inventory_df["Stock_Status"]
    == "Overstock",
    "Recommendation"
] = "Reduce Future Orders"

inventory_df.head()
```

```python
print(
    inventory_df["Stock_Status"]
    .value_counts()
)
```

```python
reorder_count = (
    inventory_df["Stock_Status"]
    == "Reorder Required"
).sum()

overstock_count = (
    inventory_df["Stock_Status"]
    == "Overstock"
).sum()

print(
    f"Reorder Required: {reorder_count}"
)

print(
    f"Overstock: {overstock_count}"
)
```

```python
inventory_output = inventory_df[
    [
        "StockCode",
        "Description",
        "Average_Daily_Demand",
        "Net_Quantity",
        "Safety_Stock",
        "Reorder_Point",
        "Inventory_Gap",
        "Stock_Status",
        "Recommendation"
    ]
]

inventory_output.to_csv(
    "../outputs/inventory_recommendations.csv",
    index=False
)

print(
    "inventory_recommendations.csv saved"
)
```
